# **CoNVO Framework for Business Understanding**

⭐ **Objective:** Drive structured thinking when understanding business context and scoping out data science projects.

⭐ **Outcome:** asking the right questions and fully scoped problem, with projects delivered successfully and on time.

## **Context (Co)** - Start with Why

**Purpose:**

Understand the situation and what triggered the request. Before diving into “how”, get clear on “why?” and “so what?”. Build enough context around the product, customer, business, and recent events to understand why this problem matters now.


Start with why.

**Questions to help gather context:**

* What triggered this request? What's happening and why are we looking at this now?
* Who are the stakeholders involved? what is there broader mission?
* What timelines are we working towards?
* Why does this matter to the business or customer?
* Who are the users, customers, or teams involved? what are there motivations, goal and expectation?
* Why are we looking at this now?
* What is happening with the product, customers, or business?
* What has changed recently?
* Where does this sit in the customer journey or funnel?
* What metrics are relevant?
* What do we already know?
* What analysis, experiments, or research have already been done?
* What do stakeholders currently believe is happening?

**Example of Context:**

*Revenue from credit products has declined over recent months, driven in part by fewer customers taking credit through the app.*


*The Commercial and CRM teams want to understand whether there is an opportunity to use CRM communications more effectively to encourage eligible customers to take credit.*

*Today, CRM campaigns can reach a broad population of eligible customers, but there is limited understanding of which customers are most likely to have a genuine need for credit, when that need might arise, and when CRM contact is most likely to influence their behaviour.*

*In short: Revenue is under pressure, credit take-up is an important commercial lever, and CRM is a potential channel through which the business can influence customer behaviour.*






## **Need (N)** - Find the real problem, not the requested solution

**Purpose:**

Define the actual problem, opportunity, or uncertainty that needs addressing. Get underneath the initial request to understand what the team really needs to know, change, or decide.



Find the real problem, not the requested solution.

**Questions to ask:**

* What problem are we actually trying to solve?
* What user or customer behaviour are we trying to understand or change?
* What don't we currently understand? what decision are we looking to make? what question(s) do we want to answer?
* What isn't working as expected?
* What evidence suggests there is a problem or opportunity?
* Who is affected?
* How significant is the problem?
* What hypotheses do we already have? do we have any evidence to support?
* What decision is currently difficult to make?
* What information are we missing?
* What happens if we do nothing?
* What would you do differently if you had the answer?

**Example of Need:**

*The business needs a better way to identify which eligible customers should receive credit-related CRM communications and when, rather than treating all eligible customers in the same way.*

*We need to understand whether customer characteristics and in-app behaviour can help identify customers who may be approaching a relevant credit need, while distinguishing between customers who would take credit anyway and those whose behaviour could genuinely be influenced by CRM.*

*This means answering questions such as:*

- *Are there observable behaviours that indicate an emerging need for credit?*
- *Which customers are more likely to take credit?*
- *When in the customer journey is that likelihood highest?*
- *Which customers are actually persuadable through CRM rather than likely to convert anyway?*
- *Are there customer groups we should avoid targeting?*
- *What CRM treatments, messages, or timings are most effective?*

*Essentially, the underlying need is to help Commercial and CRM identify where targeted CRM intervention can generate incremental credit take-up.*

## **Vision (V)** - Picture the destination before starting




**Purpose:**

Describe what the solution could look like. Imagine the desired future state before deciding on the analytical solution. Focus on what the product, marketing team, or customer should ultimately be able to do differently with the solution. It also worth considering reward vs effort. In practice, the best solution might not necessarily be the most optimal.

**Questions to ask:**

* Who would use the insight or capability?
* what would "better" look like?
* How do they intend to use it?
* When would they need it?
* What would we understand that we don't understand today?
* What would the team be able to do that they can't do today?
* What decision would the team be able to make that they can't do today?


**Example of Vision:**



*Enable Commercial and CRM to identify high-propensity customer segments for credit and surface them directly in the CRM tool for targeted campaigns.*

*The vision: CRM can target the right eligible customers, at the right time, with relevant credit propositions to drive incremental credit take-up.*

## **Outcome (O)** - Connect Solution to Action and Impact

**Purpose:**

Turn the vision into something actionable and measurable. Define what decision or action the work should enable, what behaviour you expect to change, and how you'll know whether it created value.

**Questions to ask:**

* What decision should this work inform?
* What action should someone be able to take?
* Who will take that action?
* What user or customer behaviour should change?
* What is the primary success metric?
* What guardrail metrics matter?
* What is the current baseline?
* What magnitude of change would be meaningful?
* Over what timeframe should we expect an impact?
* Which users, customers, markets, or segments should be affected?
* How will we measure whether it worked?
* Can we establish whether our action caused the outcome?
* What would make us decide not to take action?
* What does success ultimately mean for the customer or business?

**Example of Outcome:**

*Enable Commercial and CRM to make better decisions about who to target and when, increasing the effectiveness of credit CRM campaigns.*

*Success looks like:*

* *Increased incremental credit take-up from CRM.*
* *Increased incremental revenue.*
* *More efficient and relevant CRM targeting.*

*Ultimately: use data-driven targeting to generate incremental credit take-up and revenue through CRM.*
